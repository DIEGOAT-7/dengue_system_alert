```python
import pandas as pd
import os

ruta_catalogo = os.path.join('..', 'data', '01_raw', 'CNE_IDEAM.xls')

print(f"Cargando el catálogo maestro desde: {ruta_catalogo}")

try:

    catalogo_df = pd.read_excel(ruta_catalogo, engine='xlrd')
    
    print(f"¡Catálogo completo cargado! {len(catalogo_df)} estaciones en total.")
    
    
    print("\nFiltrando estaciones para Cali...")
    
    # Nos aseguramos de que las columnas sean de tipo 'str' antes de usar .str
    catalogo_df['MUNICIPIO'] = catalogo_df['MUNICIPIO'].astype(str)
    catalogo_df['ESTADO'] = catalogo_df['ESTADO'].astype(str)
    catalogo_df['CATEGORIA'] = catalogo_df['CATEGORIA'].astype(str)
    catalogo_df['TECNOLOGIA'] = catalogo_df['TECNOLOGIA'].astype(str)
    catalogo_df['CODIGO'] = catalogo_df['CODIGO'].astype(str)


    estaciones_cali = catalogo_df[
        (catalogo_df['MUNICIPIO'].str.upper() == 'CALI') &
        (catalogo_df['ESTADO'].str.upper() == 'ACTIVA')
    ].copy() # .copy() para evitar warnings

    print(f"¡Encontramos {len(estaciones_cali)} estaciones activas en Cali!")

    
    print("\n--- Desglose por Tecnología en Cali ---")
    print(estaciones_cali['TECNOLOGIA'].value_counts())

    
    # Lista para PRECIPITACIÓN (Pluviométricas + Climáticas)
    codigos_precipitacion = estaciones_cali[
        estaciones_cali['CATEGORIA'].str.contains('PLUVIOMÉTRICA|CLIMÁTICA', case=False, na=False)
    ]['CODIGO'].tolist()

    # Lista para TEMPERATURA (Solo Automáticas + Climáticas)
    codigos_temperatura = estaciones_cali[
        (estaciones_cali['TECNOLOGIA'].str.contains('AUTOMÁTICA', case=False, na=False)) &
        (estaciones_cali['CATEGORIA'].str.contains('CLIMÁTICA', case=False, na=False))
    ]['CODIGO'].tolist()

    print(f"\nCódigos encontrados para Precipitación (Pluvi + Clim): {len(codigos_precipitacion)}")
    print(f"Primeros 10: {codigos_precipitacion[:10]}")
    
    print(f"\nCódigos encontrados para Temperatura (Auto + Clim): {len(codigos_temperatura)}")
    print(codigos_temperatura)
    
    # Guardar nuestras listas filtradas
    estaciones_cali.to_csv(os.path.join('..', 'data', '03_processed', 'estaciones_cali_filtradas.csv'), index=False)
    print("\n¡Análisis local guardado en data/03_processed/!")

except FileNotFoundError:
    print(f"¡ERROR! No se encontró el archivo en {ruta_catalogo}")
    print("Asegúrate de haber movido 'CNE_IDEAM.xls' a la carpeta 'data/01_raw/'")
except Exception as e:
    print(f"Ocurrió un error al leer el Excel: {e}")
    print("Asegúrate de haber instalado 'xlrd' con 'pip3 install xlrd'")
```
